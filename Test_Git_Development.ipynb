```python
file_path = '/Volumes/workspace/default/sample_volume/medals.csv'

df = spark.read.csv(file_path, inferSchema=True, header=True)
```

```python
df.display()
```

```python
df2 = df.\
    withColumnRenamed("Rank by Total", "rank_by_total").select("Rank","Country","Gold","Bronze",'rank_by_total')
```

```python
df2.write.format("delta").mode("append").save('/Volumes/workspace/default/sample_volume/medal_delta/')
```

```python
delta_path = '/Volumes/workspace/default/sample_volume/medal_delta'

delta_df = spark.read.format('delta').load(delta_path)
delta_df.display()
```

```python
delta_df.drop('Rank').write.format('delta').mode("overwrite").save(delta_path)
```

```python
spark.read.format('delta').load(delta_path).display()
```

```python
history_df = spark.sql(f"DESCRIBE HISTORY '{delta_path}'")
display(history_df)
```
